```python
%sql
CREATE TEMP VIEW worlds_view AS SELECT playerid AS player_id, playername AS player_name, teamname AS team_name,
CASE WHEN position = 'jng' THEN 'jg' ELSE position END AS position,
COUNT(gameid) AS games_played,
ROUND(try_divide(SUM(kills) + SUM(assists), SUM(deaths)), 1)  AS kda,
2014 AS valid_from, 
CAST(NULL AS INT) AS valid_to,
TRUE AS is_current

FROM bronze_worlds_data
WHERE league = 'WLDs'
AND playerid IS NOT NULL AND playername IS NOT NULL
GROUP BY playerid, playername, teamname, position, EXTRACT(YEAR from date)
ORDER BY playername;
```

```python
%sql
SELECT * FROM worlds_view; 
```

```python
%sql
CREATE TABLE IF NOT EXISTS silver_worlds_data
(
    player_id STRING,
    player_name STRING,
    team_name STRING, 
    position STRING,
    games_played INT,
    kda DOUBLE,
    valid_from INT,
    valid_to INT,
    is_current BOOLEAN
)
```

```python
%sql 
INSERT INTO silver_worlds_data SELECT * FROM worlds_view; 
```
